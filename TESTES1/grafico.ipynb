```python
import pandas as pd
import matplotlib.pyplot as plt

x = [e for e in range(-15,21,2)]
y = [(e**2)-(5*e)+6 for e in x]
z = [e+5 for e in x]

tabela = {
    'X': x,
    'Y': y,
    'Z': z
}

grafico = pd.DataFrame(tabela)
grafico
```

```python
grafico.plot.line(x='X', alpha=1)
plt.show()
```
